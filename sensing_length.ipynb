```python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.patches as pat
import matplotlib.animation as anim
import scipy.optimize as sopt
import scipy.interpolate as strp
import scipy.integrate as sint
import scipy.sparse as spar
from copy import deepcopy


import functools as ft
import copy as cp
from tqdm import tqdm


import chart_studio.plotly as csp
import plotly as ply
import plotly.graph_objects as go
import plotly.express as px
import plotly.io as pio
pio.renderers.default = "vscode"

from IPython.display import display, HTML

ply.offline.init_notebook_mode()
display(HTML(
    '<script type="text/javascript" async src="https://cdnjs.cloudflare.com/ajax/libs/mathjax/2.7.1/MathJax.js?config=TeX-MML-AM_SVG"></script>'
))

pi = np.pi
```

```python
ns = 1001
nt = 2001
T = 40
dt = T/(nt-1)

l = 1.
ls = 0.1
l_i = np.ones(ns)*l/ns
s_node = np.insert(np.cumsum(l_i), 0, 0)
s_mid = (s_node[1:]+s_node[:-1])/2
l = np.sum(l_i)

theta_init = pi/2
kappa = np.zeros(ns)
theta = theta_init + np.insert(np.cumsum(kappa*l_i), 0, 0)
theta_mid = (theta[1:]+theta[:-1])/2

beta = 1
gamma = .2

# defined at nodes
theta_list = []
theta_list.append(theta)
# defined for each segment
theta_mid_list = []
theta_mid_list.append(theta_mid)
kappa_list = []
kappa_list.append(kappa)

```

```python
def kappa_sensed(ls):
  kappa = []
  i_shift = int(np.floor(ls/2/(l/ns)))
  for i in range(ns):
    lim_api = int(np.ceil(np.min((i + i_shift + .5, ns))))
    lim_bas = int(np.floor(np.max((i - i_shift + .5, 0))))
    kappa.append((theta[lim_api] - theta[lim_bas])/ls)
    # kappa.append((theta[lim_api] - theta[lim_bas])/((lim_api-lim_bas)*l/ns))
  return np.array(kappa)
```

```python
for t in range(nt):
  dkappa = (- beta * np.sin(theta_mid) - gamma * kappa_sensed(ls)) * dt
  kappa += dkappa
  theta = theta_init + np.insert(np.cumsum(kappa*l_i), 0, 0)
  theta_mid = (theta[1:]+theta[:-1])/2

  theta_list.append(theta)
  theta_mid_list.append(theta_mid)
  kappa_list.append(kappa)
```

```python
# Prepare for animation
frameskip = 20
frames = range(0, nt, frameskip)
x2plot = []
y2plot = []
t = []
max_x, min_x = (0,0)
max_y, min_y = (0,0)
for i in frames:
  x = sint.cumulative_trapezoid(np.sin(theta_list[i]), s_node)
  y = sint.cumulative_trapezoid(np.cos(theta_list[i]), s_node)
  x2plot.append(x)
  y2plot.append(y)
  t.append(i*dt)
  max_x, min_x = (max(max_x, np.max(x)), min(min_x, np.min(x)))
  max_y, min_y = (max(max_y, np.max(y)), min(min_y, np.min(y)))

w = max_x - min_x
h = max_y - min_y
lims = min_x - 0.1*w, max_x + 0.1*w, min_y - 0.1*h, max_y + 0.1*h

fig = go.Figure()

for it in range(len(t)):
  fig.add_trace(
    go.Scatter(
      visible=False,
      line=dict(color="blue", width=3),
      name="t = " + "{:.3f}".format(t[it]),
      x = x2plot[it],
      y = y2plot[it]
    )
  )

fig.data[0].visible=True

steps = []
for it in range(len(t)):
  step = dict(
    method="update",
    args=[{"visible": [False] * len(fig.data)},
    {"title": "Stem shape at t = " + "{:.3f}".format(t[it])}], label="{:.2f}".format(t[it])
  )
  step["args"][0]["visible"][it] = True
  steps.append(step)

sliders = [dict(
  active=0, currentvalue={"prefix": "t = "},
  pad={"t": 1},
  steps=steps
)]

fig.update_layout(width=600,height=600,
  xaxis=dict(range=[lims[0], lims[1]], autorange=False, zeroline=True),
  yaxis=dict(range=[lims[2], lims[3]], autorange=False, zeroline=False),
  title="Stem shape at t = 0" ,sliders=sliders
)
fig.update_yaxes(
  zeroline=True, zerolinewidth=1, zerolinecolor='gray',
  scaleanchor="x",
  scaleratio = 1
)

# fig.show()
# fig.write_html("sample.html")
```
